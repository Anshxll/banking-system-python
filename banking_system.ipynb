```python
import os

FILE_NAME = "accounts.txt"


class BankAccount:
    def __init__(self, account_number, name, pin, balance=0, transactions=None):
        self.account_number = account_number
        self.name = name
        self.pin = pin
        self.balance = float(balance)
        self.transactions = transactions if transactions else []

    def deposit(self, amount):
        try:
            amount = float(amount)
            if amount > 0:
                self.balance += amount
                self.transactions.append(f"Deposited {amount}")
                print("Deposit successful.")
            else:
                print("Invalid amount.")
        except:
            print("Invalid input.")

    def withdraw(self, amount):
        try:
            amount = float(amount)
            if amount > self.balance:
                print("Insufficient balance.")
            elif amount <= 0:
                print("Invalid amount.")
            else:
                self.balance -= amount
                self.transactions.append(f"Withdrawn {amount}")
                print("Withdrawal successful.")
        except:
            print("Invalid input.")

    def check_balance(self):
        print(f"Current Balance: {self.balance}")

    def show_transactions(self):
        print("Transaction History:")
        for t in self.transactions:
            print("-", t)


# ---------------- FILE HANDLING ----------------

def load_accounts():
    accounts = {}
    if not os.path.exists(FILE_NAME):
        return accounts

    with open(FILE_NAME, "r") as file:
        for line in file:
            data = line.strip().split("|")
            account_number = data[0]
            name = data[1]
            pin = data[2]
            balance = data[3]
            transactions = data[4].split(",") if data[4] else []

            accounts[account_number] = BankAccount(
                account_number, name, pin, balance, transactions
            )
    return accounts


def save_accounts(accounts):
    with open(FILE_NAME, "w") as file:
        for acc in accounts.values():
            transactions_str = ",".join(acc.transactions)
            file.write(
                f"{acc.account_number}|{acc.name}|{acc.pin}|{acc.balance}|{transactions_str}\n"
            )


# ---------------- MAIN SYSTEM ----------------

def create_account(accounts):
    account_number = input("Enter new account number: ")
    if account_number in accounts:
        print("Account already exists.")
        return

    name = input("Enter name: ")
    pin = input("Set 4-digit PIN: ")

    accounts[account_number] = BankAccount(account_number, name, pin)
    save_accounts(accounts)
    print("Account created successfully.")


def login(accounts):
    account_number = input("Enter account number: ")
    pin = input("Enter PIN: ")

    if account_number in accounts and accounts[account_number].pin == pin:
        print("Login successful.")
        return accounts[account_number]
    else:
        print("Invalid credentials.")
        return None


def admin_panel(accounts):
    print("\n--- ADMIN PANEL ---")
    print("Total Accounts:", len(accounts))
    for acc in accounts.values():
        print(f"{acc.account_number} | {acc.name} | Balance: {acc.balance}")


def main():
    accounts = load_accounts()

    while True:
        print("\n1. Create Account")
        print("2. Login")
        print("3. Admin Panel")
        print("4. Exit")

        choice = input("Choose option: ")

        if choice == "1":
            create_account(accounts)

        elif choice == "2":
            account = login(accounts)
            if account:
                while True:
                    print("\n1. Deposit")
                    print("2. Withdraw")
                    print("3. Check Balance")
                    print("4. Transaction History")
                    print("5. Logout")

                    option = input("Choose option: ")

                    if option == "1":
                        amount = input("Enter amount: ")
                        account.deposit(amount)
                        save_accounts(accounts)

                    elif option == "2":
                        amount = input("Enter amount: ")
                        account.withdraw(amount)
                        save_accounts(accounts)

                    elif option == "3":
                        account.check_balance()

                    elif option == "4":
                        account.show_transactions()

                    elif option == "5":
                        break

                    else:
                        print("Invalid option.")

        elif choice == "3":
            admin_panel(accounts)

        elif choice == "4":
            save_accounts(accounts)
            print("Thank you for using the bank system.")
            break

        else:
            print("Invalid choice.")


if __name__ == "__main__":
    main()
```
